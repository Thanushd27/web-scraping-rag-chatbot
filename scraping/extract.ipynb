```python
file_path = 'pages_name.txt'

lists = []

with open(file_path, 'r') as file:
    for line in file:
        url = line.strip()
        if url:
            lists.append(url)

print(lists)
```

```python
from bs4 import BeautifulSoup
import re
from docx import Document
import cloudscraper


def parse_html(html_content):
    soup = BeautifulSoup(html_content, 'html.parser')
    return soup.get_text()


def fetch_webpage(url):

    from selenium import webdriver
    from selenium.webdriver.chrome.options import Options
    import time

    options = Options()

    # DO NOT use headless
    # options.add_argument("--headless")

    options.add_argument("--disable-blink-features=AutomationControlled")

    driver = webdriver.Chrome(options=options)

    driver.get(url)

    time.sleep(8)

    html = driver.page_source

    driver.quit()

    return html


def clean_text(text):
    text = ' '.join(text.split())
    text = re.sub(r'http\S+', '', text)
    text = re.sub(r'[^a-zA-Z0-9\s.,!?\'"]+', '', text)

    return text


def write_to_docx(text_content, output_filename):

    doc = Document()

    doc.add_heading('Tesla Website Data', level=1)

    doc.add_paragraph(text_content)

    doc.save(output_filename)
```

```python
for i, url in enumerate(lists):

    html_content = fetch_webpage(url)

    page_text = parse_html(html_content)

    final_text = clean_text(page_text)

    output_filename = f"tesla_page_{i+1}.docx"

    write_to_docx(final_text, output_filename)

    print("Saved:", output_filename)
```
